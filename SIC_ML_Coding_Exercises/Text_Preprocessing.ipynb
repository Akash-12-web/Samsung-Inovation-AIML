```python
# Tokenization with spacy
import spacy
nlp = spacy.load('en_core_web_sm')

doc = nlp(text)
```

```python
doc
```

```python
# sentence tokenization
for i, sent in enumerate(doc.sents,start=1):
  print(f'{i} :{sent}')
```

```python
# word tokenize
for token in list((doc.sents))[0]:
  print(token.text, end = ' | ')
```

```python
# Noise removal & normalization
clean_tokens = []
for token in doc:
  if token.is_alpha:
    clean_tokens.append(token.text.lower())
```

```python
clean_tokens
```

```python
# stopwords removal
filtered_tokens = [token for token in clean_tokens if not nlp.vocab[token].is_stop]
```

```python
filtered_tokens
```

```python
#lemmmatization
lemmas  = []
for token in doc:
  if token.is_alpha and not token.is_stop:
    lemmas.append(token.lemma_.lower())
```

```python
lemmas
```

```python
from xml.sax.saxutils import prepare_input_source
# pre-processing pipeline/function
def preprocess_spacy(text):
  doc = nlp(text)

  preprocess_tokens = []
  for tokens in doc:
    if tokens.is_alpha and not  tokens.is_stop:
      preprocess_tokens.append(tokens.lemma_.lower())
  return preprocess_tokens
```
